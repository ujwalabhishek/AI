## (1.2) Introduction to Image Generation
The field of AI is becoming more and more popular everyday. This is, in part, thanks to the capabilities of generative AI models such as ChatGPT (for language), Stable Diffusion (for images), and so on.

You are already familiar with prompt engineering in the case of language models such as ChatGPT. You might have noticed that language models like ChatGPT require a structured description of what you want from it. You can use chain-of-thought methods along with few-shot examples to align your language model instance with what you want.

Image models are slightly different because they are not trained on the same data as language models. Image models are primarily trained on captioned images, so the models pay more attention to keywords rather than elaborate descriptions.

Consider this image:


dog.png

You'd be surprised to learn that this image is not made by a human being. In fact, this image has been generated using Stable Diffusion, which is an image generation model.

What's interesting is the prompt that was used to generate this image.

Here's the prompt:
> *Acrylic painting of a dog, shot from far, green meadow, classical art, impressionism, loose brush strokes, bright painting, Outdoors*

You can see that this is quite different from a typical ChatGPT prompt. This image prompt is essentially a combination of the following elements:
- Foreground
- Background
- Descriptors

Foreground: The main subject in your image - *Acrylic painting of a dog*

Background: The background in your image - *green meadow*

Descriptors: Terms that describe the nature and style of your image - *classical art*, *impressionism*, *loose brush strokes*, and so on

These components of an image prompt can be populated by **keywords** that the image model understands. Recall some prompts that you may have written for ChatGPT. What are some differences between image prompts and text prompts?

In text prompts, with the help of roles, few shot examples, and chain of thoughts reasoning, we tell the model *how* to solve a problem. Whereas in case of image prompts, we need to describe *what* the image should look like in great detail.

Before we move on to important keywords associated with image model prompting, let's first take a look at the motivations behind and the developmental history of image generation models.

## (1.4) History
Image generation is not new. The first successful instance of reliable generation of new images was done using variational autoencoders or VAEs. At the same time, generative adversarial networks or GANs were also used for creating images.

Though the biggest breakthrough in image generation has happened only recently with the advent of diffusion-based models such as Stable Diffusion.

Here's a simple overview of how these models work:

**Variational Autoencoders**

variational autoencoder.png

Variational Autoencoders (VAE) are neural networks that **learn** encoding parameters by training on the raw and encoded versions of images.
Variational autoencoders incorporate slight random disturbances in the decoding process which then leads to similar but completely new images.



**Generative Adversarial Networks**

GANs.png


Generative Adversarial Networks (GANs) are made up of two networks. One network tries to produce new images. The other tries to identify whether the images created are new or real. The GAN is then trained till a point where the classifier cannot reliably identify the images coming from the generator anymore. This way, sufficiently realistic new images can be created.


**Diffusion models**

diffusion.png


Diffusion models draw inspiration from various architectures such as VAEs, transformers, and so on. In short, diffusion models work by adding noise to an image in steps, while training the network to identify how much noise was added. By introducing randomness in the network, like in the case of VAEs, and by accurately removing noise, diffusion models are able to produce new images.

One thing to note is the diffusion models are multimodal models. That is, they are trained on data of different types, combined. For example, diffusion models are trained on captioned images. This is how they are able to convert text into new images.

We will now define a function that uses the official image generation method from OpenAI with the following parameters:
- *prompt*: This is the text input prompt that will be used to generate new images
- *n*: This is the number of images to generate per prompt

For this demonstration, we will use SD 2.1, which was trained on 768x768 images, and is generally better than SD 1.5, which was trained on 512x512 images.

You can read more about this model on their official [release blog post](https://stability.ai/blog/stablediffusion2-1-release7-dec-2022) or check out their [model card on HF](https://huggingface.co/stabilityai/stable-diffusion-2-1).

We will next import the Stable Diffusion pipeline that allows us to put all these pieces together, that is the model, the diffuser, the GPU, and other components.


For the purposes of this demonstration, we are downloading StableDiffusion provided by StabilityAI on Hugging Face Hub.

```python
# Import the Stable Diffusion pipeline method and set up the pipeline
# Note: You can change the model here if you want
import torch
from diffusers import StableDiffusionPipeline

model_name = 'stabilityai/stable-diffusion-2-1'
pipe = StableDiffusionPipeline.from_pretrained(model_name,
                                               torch_dtype = torch.float16,
                                               variant = 'fp16')

# Convert the pipe to the Cuda language that is used by the Nvidia GPUs
pipe = pipe.to('cuda')
```

There are many schedulers that have been used to train Stable Diffusion models such as the PNDM scheduler, DDIM scheduler, the LMSD scheduler, and so on. While selection of schedulers also plays an important role in how image generation is done by the SD model, for this demonstration, we will restrict ourselves to the default PNDM scheduler.

We will now define a function that specifies relevant parameters associated with the SD 2.1 model that control the nature of the operations being performed including parameters such as the size of the output image, the number of images to output, and so on.

Here is a description of each parameter:
- *prompt*: This is the text input that is used to generate a new image
- *negative_prompt*: This is the text input that is used to suppress parts of the output image
- *num_inference_steps*: This is the number of inference steps that SD executes before returning an output image
- *guidance_scale*: This is a value that controls how closely generated images follow the input prompts
- *height*: This is the height of the output image in number of pixels
- *width*: This is the width of the output image in number of pixels
- *num_images_per_prompt*: This is the number of images to be generated for a single input prompt
- *generator*: This is used to generate pseudorandom numbers
- *seed*: The parameter used to generate a random vector from which to produce an image

Before we move ahead with creating custom prompting functions for SD, let's fix some parameters:
- *num_inference_steps* = 45
  - Lower number of inference steps produces worse images but takes less time whereas a higher number of inference steps produces better images but takes more time
- *guidance_scale* = 7
- *height* = *width* = 768
  - The original model was trained on 768x768 images but we can change this value
- *num_images_per_prompt* = 1
  - For the purpose of this demonstration, we will generate only one image per prompt, but in practicem we can generate many variations for the same prompt

```python
import random
import math

def gen_sd_img(prompt, negative_prompt = None, seed = 0):
  rnum = math.floor(random.random() * 1000000000)
  fseed = rnum
  if(seed != 'random'):
    fseed = seed
  image = pipe(prompt = prompt,
               negative_prompt = negative_prompt,
               num_inference_steps = 45,
               guidance_scale = 7,
               height = 768,
               width = 768,
               num_images_per_prompt = 1,
               generator = torch.Generator('cuda').manual_seed(fseed)).images[0]

  return image
```

## (4.4) Prompting Stable Diffusion using Code

We will now use the prompting function to look at various generated images. We will incrementally improve our prompt engineering techqinues as we use the custom function.

```python
# Use the custom function to generate an image
gen_sd_img(prompt = 'submarine',
           negative_prompt = None,
           seed = 358870059)
```

You might notice that the output doesn't seem very relevant except for a reference to the ocean. Moreover, the human faces have not come out well. And there's some gibberish in the image as well.

This is where negative prompting comes into play. Negative prompting is used in Stable Diffusion to tell the model to suppress its focus on certain keywords.

In this case, we want the model to place less importance on human faces and text, so we can add these keywords to the negative prompt.

```python
# Use the custom function to generate an image with the same prompt along with a relevant negative prompt
gen_sd_img(prompt = 'submarine',
           negative_prompt = 'text, signature, weird face, incorrect face, anatomically incorrect face',
           seed = 358870059)
```

You can see that this time, the model has produced a much better and more relevant image. This is the importance of negative prompting in Stable Diffusion.

After much research, experts agree on some common keywords that may be included as negative prompt keywords. Her are some of them:
- *text*
- *signature*
- *weird face*
- *incorrect face*
- *anatomically incorrect face*
- *out of frame*
- *duplicate*
- *blurry*
- *low quality*
- *out of focus*

This list is by no means complete. As research improves, experts might come up with better negative prompting techniques.

One thing to note is that adding unnecessary negative prompts confuses the image model and may lead to worse images.

```python
# Use the custom function to generate an image with the same prompt along with an irrelevant negative prompt
gen_sd_img(prompt = 'submarine',
           negative_prompt = 'text, signature, weird face, incorrect face, anatomically incorrect face, weird body parts, clouds, bad machine parts',
           seed = 358870059)
```

You can see that the generated image is not as good as the previous one. The image is back to showing human bodies and there isn't even a submarine in the image. Basically, the size and contents of the negative prompt should be exactly what is required, not too less and not too much.

So, let's regenerate our original image with the proper negative prompt.

```python
# Use the custom function to generate an image with the same prompt along with a relevant negative prompt
gen_sd_img(prompt = 'submarine',
           negative_prompt = 'text, signature, weird face, incorrect face, anatomically incorrect face',
           seed = 358870059)
```
