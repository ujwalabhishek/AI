## 1.Semantic Search

Search, or the more technical term ‘information retrieval’, is the process of retrieving relevant information from a large collection of data. Semantic search can be considered to be a subset of information retrieval that seeks to improve search accuracy by understanding the user's intent and the contextual meaning of terms by applying the principles of embedding space. Semantic search uses vector search and machine learning techniques to return results that aim to match a user’s query even when there are no word matches. Let’s hear more on this from Akshay in the next video.

<div class="MuiBox-root css-1bi8ut6"><div class="text_component" data-testid="text-component"><p style="text-align: justify;">As mentioned in the video, semantic search focuses on the semantic understanding of the query and the documents by generating and comparing vector embeddings. In the previous segments, you were introduced to the concept of word embeddings that can be used to represent words as vectors in a high-dimensional space. These embeddings can be used to capture the semantic meaning of words and measure the similarity between words or documents. In semantic search ( sometimes also known as dense retrieval), the objective is to extract the document that matches closely with the user’s query. Semantic search systems go beyond simple keyword matching and take into consideration the context, semantics and conceptual relationships between words to match a user query with the corresponding content. As a result, a semantic search system can understand the intent and meaning behind a user's query and match it with relevant documents even if the query and the documents use different words or phrasing. After representing the pieces of text or documents in a high-dimensional space, the vector embeddings of the query and the documents can be compared using a distance metric such as cosine similarity. The search problem is now converted to a nearest neighbour method to find the phrase that closely matches with the vector embeddings of the query phrase and vice versa, as illustrated in the diagram below. </p><p><br></p><p><img src="https://d35ev2v1xsdze0.cloudfront.net/1a5a89c5-efbe-4d87-9636-a4425bf126c1-A.png" style="width: 706px;" class="fr-fic fr-dib" etag="&quot;4218483b6b33c3b087c4d15de2e49436&quot;" serversideencryption="AES256" location="https://d35ev2v1xsdze0.cloudfront.net/1a5a89c5-efbe-4d87-9636-a4425bf126c1-A.png" key="1a5a89c5-efbe-4d87-9636-a4425bf126c1-A.png" bucket="upgrad-prod" originalname="A.png" fieldname="upload" contenttype="image/png" mimetype="image/png" size="128314" encoding="7bit" acl="public-read"></p><p style="text-align: justify;">With the advent of transformers and the rise of LLM-enabled text embeddings, it has become possible to capture the semantic value of words and phrases beyond their surface-level syntax or spelling. </p><p><br></p></div></div>

The main difference between semantic search and dense retrieval lies in their underlying approaches and goals. Semantic search emphasises understanding the meaning and intent of the query and documents, whereas dense retrieval focuses on mapping text into a vector space to efficiently compute similarities and retrieve relevant content. ## 3.Document Ingestion

Once the text embedding model has been set up, the next major component is to deal with the documents for which the embeddings need to be generated. Documents vary widely in terms of their format, the way the information is presented within each document and the number of pages within each document. It is often not practical to embed entire documents as a single vector, particularly when dealing with long documents such as books or research papers. This is why chucking strategies are important. In the video below, Akshay will go over the various chunking strategies.

<div class="text_component" data-testid="text-component"><p style="text-align: justify;">As mentioned in the video, a chunking strategy is crucial to determine the effectiveness of a semantic search application. A chunk is a piece of text that is representative of the information contained within that section of the text. A chunking strategy involves dividing a large document into smaller, more manageable chunks for embedding, as illustrated in the diagram below.</p><p><br></p><p><img src="https://d35ev2v1xsdze0.cloudfront.net/3de1f650-1456-446c-ae78-4671657f7986-A.png" style="width: 702px;" class="fr-fic fr-dib" etag="&quot;db5ae36070011c7c73cc82b652b489aa&quot;" serversideencryption="AES256" location="https://d35ev2v1xsdze0.cloudfront.net/3de1f650-1456-446c-ae78-4671657f7986-A.png" key="3de1f650-1456-446c-ae78-4671657f7986-A.png" bucket="upgrad-prod" originalname="A.png" fieldname="upload" contenttype="image/png" mimetype="image/png" size="71836" encoding="7bit" acl="public-read"></p><p><br></p><p dir="ltr" style="text-align: justify;"><span style="font-size: 14px;">There can be various factors that influence the choice of the chunking strategy:</span></p><ul><li aria-level="1" dir="ltr" style="font-size: 14px;"><p dir="ltr" role="presentation" style="text-align: justify;">The embedding model and its optimal chunk size</p></li><li aria-level="1" dir="ltr" style="font-size: 14px;"><p dir="ltr" role="presentation" style="text-align: justify;">Nature of the context</p></li><li aria-level="1" dir="ltr" style="font-size: 14px;"><p dir="ltr" role="presentation" style="text-align: justify;">Downstream task</p></li></ul><p dir="ltr" style="text-align: justify;"><span style="font-size: 14px;"><br></span></p><p dir="ltr" style="text-align: justify;"><span style="font-size: 14px;">Most embedding models have a context window limit. A text embedding model represents words as vectors by extracting the context from the surrounding words or text. Oftentimes, the context window size of the text embedding models needs to be considered before any document chunking strategy is taken into consideration. ## Overlapping Chunking

In the previous segment, you looked at three chunking strategies: the <strong>fixed window strategy</strong>, <strong>the max token window</strong> <strong>strategy&nbsp;</strong>and <strong>natural delimiter based strategy</strong>. Each of these chunking strategies have their own set of pros and cons when used to embed information.&nbsp;</span>

In the video below, Akshay will explain the principles of another chunking strategy called the <strong>overlapping chunking strategy</strong>

<div class="MuiBox-root css-1bi8ut6"><div class="text_component" data-testid="text-component"><p style="text-align: justify;">As mentioned in the video, the overlapping window chunking strategy uses a specified number of tokens to overlap with the subsequent chunks. The purpose of the overlapping window chunking strategy is to maintain contextual continuity while breaking down large pieces of text into smaller segments. This approach helps ensure that the semantic context is not lost between chunks and instead it's shared among the various chunks. This introduces redundancy in the chunks but at the same time preserves the information/context. The image below illustrates the overlapping window chunking strategy. </p><p><img src="https://d35ev2v1xsdze0.cloudfront.net/ca74981e-1c48-47a1-b68a-3727dd5994f5-A.png" style="width: 702px;" class="fr-fic fr-dib" etag="&quot;88a725aac80268ed635d88b7c324a300&quot;" serversideencryption="AES256" location="https://d35ev2v1xsdze0.cloudfront.net/ca74981e-1c48-47a1-b68a-3727dd5994f5-A.png" key="ca74981e-1c48-47a1-b68a-3727dd5994f5-A.png" bucket="upgrad-prod" originalname="A.png" fieldname="upload" contenttype="image/png" mimetype="image/png" size="680289" encoding="7bit" acl="public-read"></p><p style="text-align: justify;">The usual method to implement the overlapping window chunking strategy, is to split the text by a fixed number of tokens with a percentage overlap to the subsequent chunk.&nbsp;</p><p style="text-align: justify;"><br></p><p style="text-align: justify;">The overlapping window chunking strategy is a useful technique for breaking down large pieces of text into smaller segments while maintaining contextual continuity and is generally helpful when&nbsp;semantic meaning is captured at the sentence level but can be extended beyond sentence-level representation as well.
