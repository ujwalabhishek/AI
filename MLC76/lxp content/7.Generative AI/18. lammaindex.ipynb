# Core Components of LlamaIndex

<div class="text_component" data-testid="online-editor-content"><p style="text-align: justify;"><strong></strong><br><br>LlamaIndex is a framework designed to facilitate the interaction between large language models (LLMs) and structured data for search and retrieval applications. The core components of LlamaIndex are as follows:</p><p style="text-align: justify;">&nbsp;</p><p dir="ltr" style="text-align: justify;">Index:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">In LlamaIndex, data is handled in units called documents. These documents can encompass various types of text data.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Each document is divided into smaller units known as <em>nodes</em>. Think of nodes as the equivalent of broken down portions, such as paragraphs or sections, of a lengthy article.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The indexing process involves creating nodes and associating embeddings with the data within each node.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The basic index consists of several JSON objects, including <em>document storage</em>, <em>index storage</em> and <em>vector storage</em>.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;"><em>Document storage</em>&nbsp;contains ingested node objects. These can be stored in memory, MongoDB or Redis.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;"><em>Index storage</em>&nbsp;holds lightweight index metadata, and MongoDB or Redis can be used as a storage backend.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;"><em>Vector storage</em>&nbsp;stores the embedding vectors of ingested document chunks. By default, LlamaIndex uses an in-memory vector store, but it also supports various external vector sources such as Wave, Pinecone and Quadrant.</p></li></ul><p dir="ltr" style="text-align: justify;">&nbsp;</p><p dir="ltr" style="text-align: justify;">Retriever:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The retriever is a crucial component in LlamaIndex responsible for fetching the most relevant data nodes based on a user's query.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">When a user submits a query, the retriever's role is to retrieve the most pertinent nodes from the indexed data.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The number of relevant nodes returned can be controlled by adjusting parameters, similar to the <em>top-k</em> setting.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Different types of indexes may have different retrievers, and users can configure these retrievers for specific requirements.</p></li></ul><p style="text-align: justify;">&nbsp;</p><p dir="ltr" style="text-align: justify;">Response Synthesizer:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Once the retriever has completed its job and fetched relevant nodes, the response synthesizer takes over.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The response synthesizer processes the user's query and the retrieved nodes to generate an appropriate answer.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">There are various modes of operation for the synthesizer, including "refine," "accumulate," "compact," "accumulate and compact," "summarize" and "simple summarize." These modes determine how the synthesizer interacts with the data nodes to produce responses.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The chosen mode can affect how the synthesizer queries an LLM and how it presents the response to the user.</p></li></ul><p style="text-align: justify;">&nbsp;</p><p dir="ltr" style="text-align: justify;">Query Engine:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The query engine is a component that coordinates with all the other components to provide a response to the user's query.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">It acts as the central component that brings together the index, retriever and response synthesizer, ensuring they work together seamlessly.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The query engine interprets the user's query, initiates the retrieval process and generates a response that is then presented to the user.</p></li></ul><p style="text-align: justify;">&nbsp;</p><p dir="ltr" style="text-align: justify;">The core components of LlamaIndex work in tandem to make it efficient and straightforward to access and present relevant information based on user queries. This framework provides flexibility and customization options, enabling users to tailor their search and retrieval systems to meet specific needs.</p></div>

# Indices - Different Types

<div class="text_component" data-testid="online-editor-content"><p style="text-align: justify;"><strong></strong><br>&nbsp;</p><p dir="ltr" style="text-align: justify;">LamaIndex is a system for managing and retrieving text data efficiently. It offers multiple types of indices to cater to different text processing needs, from summarisation to keyword-based retrieval.</p><p style="text-align: justify;">LamaIndex is not a traditional SQL index but a custom text indexing mechanism. It employs different types of indices to organise, store and retrieve information from text documents. Each index serves a specific purpose, such as efficient retrieval, summarisation, or keyword-based searching. These indices are essential tools for building and querying QA systems, and they are especially valuable for applications involving large text collections.</p><p dir="ltr" style="text-align: justify;">Let’s take a look at these indices one by one.</p><p style="text-align: justify;">Vector Store Index:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The Vector Store Index is a component of the LamaIndex and is used to efficiently store and retrieve information from large collections of text documents. It does not work like traditional SQL database indexes; instead, it divides documents into smaller chunks and computes an embedding for each chunk to facilitate quick retrieval.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Documents are divided into smaller portions called <em>nodes</em>. Each node corresponds to a chunk of text from the document.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">For each node, an embedding is computed. Embeddings are numerical representations of the text content, allowing for similarity comparisons.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The LamaIndex converts document objects into node objects and stores each node and its corresponding embedding in a vector store.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">When you want to find information in your documents, you query the Vector Store Index. It involves fetching the top K similar nodes based on embedding similarity.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Once the similar nodes are retrieved, they are passed to the <strong>Response Synthesis Module</strong>. This module takes these nodes and generates responses to your queries, often using AI models such as GPT-4.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">This type of index is useful when you want to retrieve text data efficiently and generate responses for questions or queries.</p></li></ul><p style="text-align: justify;">List Index:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The List Index is another part of the LamaIndex system that organises and retrieves text data. It is particularly useful for summarising.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Data is segmented into distinct sets of nodes and linked together in a sequence. This sequence can be thought of as an ordered list of nodes.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">During querying, you can retrieve nodes directly or based on embedding similarity using the <em>similar_to </em>parameter. This parameter allows you to retrieve nodes that are the most similar to a reference node.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The List Index can be effectively used for summarisation. It goes through all the available nodes to generate a summary of the given document.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The service context parameter <em>response_mode </em>can be set to "tree_summarize" to facilitate summarisation.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">This index is ideal when you need to generate summaries or structured responses from your text documents.</p></li></ul><p style="text-align: justify;">Keyword Table Index:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">The Keyword Table Index focuses on extracting and managing keywords within text documents to facilitate efficient retrieval of relevant information.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">It extracts keywords from each node in a document and creates a mapping from these keywords to the nodes that contain them.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">When a query includes specific keywords, this index filters nodes that contain those keywords. These filtered nodes are then used to generate responses.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">This index is particularly efficient for quickly retrieving nodes that contain specific keywords. It helps connect queries with relevant content in the document.</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">It is valuable when you want to quickly locate information related to specific keywords or topics within your text documents.</p></li></ul></div>
