# How Large Language Models Are Trained

In this segment, we will discuss how large language models (LLMs) are trained.</p><p dir="ltr" style="text-align: justify;">LLMs are trained through a two-stage training process which involves the following steps:&nbsp;</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Pre-training</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Fine-tuning.<br><img class="image-editor" maxwidth="100%" src="https://d35ev2v1xsdze0.cloudfront.net/82256c9c-ddac-43ec-9b5f-2b5c8c3cd35f-A.png"></p><p dir="ltr" style="text-align: justify;">Let’s now look at these two stages in more detail.</p><p dir="ltr" style="text-align: justify;"><strong>Pretraining</strong></p><p dir="ltr" style="text-align: justify;">During the pre-training stage, the LLM is exposed to a vast and diverse set of unlabelled text data. This corpus can include anything from books and articles to websites and other forms of written communication. Pre-training requires significant computational power and time because the model is being trained on a massive dataset. Besides, due the complex architecture of LLMs, especially the self-attention modules that power the LLMs’ ability to determine complex relationships between different parts of the text, we need to perform training for multiple epochs. Here, one epoch refers to one passthrough of the whole data set.</p><p dir="ltr" style="text-align: justify;">The pretraining process of Large Language Models (LLMs) begins with the model in an untrained state where the initial parameters, known as weights, are set randomly.<br><img class="image-editor" maxwidth="100%" src="https://d35ev2v1xsdze0.cloudfront.net/7d7ebed4-164f-4e3f-882c-32cfa8c7becd-A.png"><br>The fundamental architecture of an LLM is designed to process language in discrete pieces called "tokens." These tokens can be words, parts of words, or even punctuation marks—essentially the atoms of written language. The task for which the LLM is being pretrained is called "next token prediction." This means that the model is presented with a string of tokens and must predict the subsequent token in the sequence.&nbsp;</p><p style="text-align: justify;">For example, given the phrase "The cat sat on the," the model might predict "mat" as the next token. This process is iterative and involves presenting the model with large amounts of text so that it can learn to anticipate what comes next in any given snippet of language.</p><p style="text-align: justify;">This training method is self-supervised. To understand what self-supervised means, let us first recall supervised and unsupervised training.&nbsp;</p></li><li aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;"><strong>Supervised training</strong>: In supervised training, you train the ML model using labelled data. These labels supervise the learning process by telling the model the correct outcome in a data point. For example, an image might be labelled as ‘cat’ or ‘dog’ based on the animal being captured in that image.</p></li><li aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;"><strong>Unsupervised training</strong>: In unsupervised training, the ML model analyses unlabelled data sets. These models can identify patterns in data without guidance from a human being. In these tasks, there is no obvious target variable whose value we are trying to predict. For example, an ML model called K-means clustering divides the data set into clusters with K members each. These clusters are created based on the similarity of the features of its members.</p><p dir="ltr" role="presentation" style="text-align: justify;"><br>Self-supervised learning leverages the text itself to provide the supervision. Since the correct next token is part of the input data, the model uses the input to teach itself without the need for external labels. This process is performed on a massive data set. </pre><p style="text-align: justify;">Through this process, the LLM learns the structure and rules of language, absorbing not only grammar and syntax but also general knowledge that is implicitly contained within the text. However, because of the vastness of language and the breadth of knowledge to be learned, this phase is both computationally expensive and time-consuming.<br>&nbsp;</p></li></ul><p style="text-align: justify;"><strong>Fine-Tuning</strong></p><p style="text-align: justify;">After pre-training, the second step in training LLMs is fine-tuning. In this stage, the model is trained further, but now on a much smaller, more focused dataset that may be labelled or unlabelled, depending on the desired task. This dataset is tailored to the specific application for which the LLM will be used, such as medical language for healthcare applications or legal jargon for legal applications. Fine-tuning allows the model to adjust its pre-trained parameters to better understand and produce text that is specific to its intended use case. This stage is less resource-intensive than pre-training, both in terms of data required and computational resources, due to the more narrow scope of learning.</p><p style="text-align: justify;">Fine-tuning is the process of refining a pre-existing foundation model to enhance its performance for specific tasks or knowledge areas. For instance, OpenAI's GPT-3, a general-purpose language model, underwent fine-tuning to create ChatGPT, a variant more adept at conversational responses.</p><p style="text-align: justify;">Fine-tuning can employ both self-supervised and supervised learning techniques. Self-supervised learning continues the work of the pre-training phase, where the model still learns from unlabelled data by predicting next tokens or filling in blanks. Supervised learning involves labelled data, where the model is given explicit examples of input-output pairs to learn from. For instance, to fine-tune a model for customer service, it might be trained on transcripts of customer interactions where the correct responses are labelled.</p><p style="text-align: justify;">The goals of fine-tuning are two-fold. Firstly, it may aim to increase the model's knowledge about a specialised topic, making it more informed and better able to understand queries about that topic. Secondly, it can alter the model's behaviour to make it more suitable for specific applications, such as training it to follow instructions or maintain a conversation.</p><p style="text-align: justify;">You have now learned about how large language models are trained. # Why Fine-Tune?

So far in this session, you have learned about the general process for training large language models. You have also seen that fine-tuning is an integral part of the training process of LLMs and provides vital additional capabilities to our LLM as compared to the foundation model.</p><p style="text-align: justify;">You have previously interacted with LLMs in various different ways such as by prompting them, by creating applications using frameworks like Langchain and LlamaIndex, using retrieval augmented generation and so on. These approaches allow you to customise the behaviour of an LLM in different ways, but fine-tuning a model lets you achieve the same results with better consistency.</p><p style="text-align: justify;">Therefore, to answer the question “Why fine-tune?” we need to approach it from two different angles:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Why fine-tune a foundation model?</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Prompt engineering vs. Fine-tuning</p></li></ul><p dir="ltr" style="text-align: justify;">In this segment, we will consider these questions and also talk about the process of fine-tuning in more detail.</p><p dir="ltr" style="text-align: justify;"><strong>Why fine-tune a foundation model?</strong></p><p dir="ltr" style="text-align: justify;">In the case of a foundation model (the outcome of the pretraining stage), the task for which it has been trained is “next token prediction”. # How to Fine-Tune LLMs

We can take three approaches to perform fine-tuning on LLMs:</p><ul><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Self-supervised learning</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Supervised learning</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Reinforcement learning with human feedback</p></li></ul><p dir="ltr" style="text-align: justify;">Let’s look at these three approaches individually.</p><p dir="ltr" style="text-align: justify;"><br><strong>Self-Supervised Learning</strong></p><p dir="ltr" style="text-align: justify;">Self-supervised learning focuses on the inherent structure of the data without requiring explicit labels. The training process looks very similar to the pre-training process except for the fact that the training data set contains some specialised knowledge.&nbsp;<br><img class="image-editor" maxwidth="100%" src="https://d35ev2v1xsdze0.cloudfront.net/fda1f536-70e5-4f54-9b53-50a846bf94c9-A.png"></p><p dir="ltr" style="text-align: justify;">The LLM continues to learn the task of next-token prediction as it did during pretraining, but now, it does so in a context designed to refine its capabilities in a specific area. For instance, by feeding the model a corpus of text from a certain author or genre, the model can pick up on and generate text with similar stylistic features. Another example as we saw earlier could be to gain knowledge in a specialised field such as endocrinology in medicine.&nbsp;</p><p dir="ltr" style="text-align: justify;"><br><strong>Supervised Learning</strong></p><p dir="ltr" style="text-align: justify;">Supervised learning, on the other hand, involves training on input-output pairs, which act as explicit examples for the model to learn from. This method relies heavily on the curation of a high-quality training dataset that contains examples closely aligned with the desired task. An example of this data set could be pairs of instructions and responses as given in the image below.&nbsp;<br><img class="image-editor" maxwidth="100%" src="https://d35ev2v1xsdze0.cloudfront.net/b286e86e-35c2-49d2-b397-2269e3022add-A.png"><br>Once the data set has been curated, we need to convert it into pairs of inputs and outputs by injecting the instruction into a prompt and using the response as the output. This application of supervised learning could enable the LLM to fine-tune its abilities to respond accurately to instructions.<br><br><br><strong>Reinforcement Learning</strong></p><p dir="ltr" style="text-align: justify;">Reinforcement learning introduces a reward model to guide the fine-tuning process. In this approach, the LLM learns to generate responses that are not only contextually and grammatically correct but also aligned with human preferences. The reward model typically involves human feedback, which assesses the quality and relevance of the model's responses. Over time, the LLM is trained to optimise its outputs to maximise this feedback, essentially learning to produce responses that are more likely to be favoured by users.</p><p style="text-align: justify;">&nbsp;</p><p style="text-align: justify;">After considering the three different types of fine-tuning, let’s look at the steps that we need to undertake every time we perform fine-tuning.</p><ol><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Choose fine-tuning task</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Prepare training data set</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Choose a base model</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Perform fine-tuning</p></li><li dir="ltr" aria-level="1"><p dir="ltr" role="presentation" style="text-align: justify;">Evaluate model performance</p></li></ol><p dir="ltr" style="text-align: justify;">Before moving forward, answer the following questions to check your understanding.</p></div>
